```python
import numpy as np

data = np.loadtxt('./data-dz2.txt')
```

```python
print(round(data.mean(), 4))
mean = data.mean()
lamb_1 = 1/mean
print(round(lamb_1,4))
```

```python
moment_2 = 0
for i in range(len(data)):
    moment_2 += data[i]*data[i]
moment_2 /= len(data)
print(round(moment_2,4))
lamb_2=np.sqrt(2/moment_2)
print(round(lamb_2,4))
```
